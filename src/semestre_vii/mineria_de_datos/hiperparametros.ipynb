```python
import numpy as np
import pandas as pd
from scipy.stats import randint, uniform
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score, accuracy_score, confusion_matrix
```

```python
semilla = 42

X,y = make_classification(
    n_samples=5_000,
    n_features=20,
    n_informative=5,
    n_redundant=3,
    weights=[0.97, 0.03],
    random_state= semilla
)
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify= y, random_state=semilla
)


pipeline = Pipeline([
    ("escalador", StandardScaler()),
    ('selector', SelectKBest(score_func=f_classif)),
    ('modelo', RandomForestClassifier(class_weight='balanced', random_state=semilla))
])



hiperparametros = {
    'selector__k': randint(4, 15),
    'modelo__n_estimators': randint(50, 250),
    'modelo__max_depth': [4,6,8,12,None],
    'modelo__min_samples_split': randint(2, 15),
}

cv_estratificado = StratifiedKFold(n_splits=5, shuffle=True, random_state=semilla)


busqueda_hiperparametros = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=hiperparametros,
    n_iter=25,
    cv=cv_estratificado,
    scoring='roc_auc',
    n_jobs=-1,
    random_state=semilla
)

busqueda_hiperparametros.fit(x_train, y_train)
```

```python
print("Mejores hiperparámetros encontrados:")
print(round(busqueda_hiperparametros.best_score_, 4))
```

```python
pd.DataFrame(busqueda_hiperparametros.cv_results_)
```

```python
for  parametro, valor in busqueda_hiperparametros.best_params_.items():
    print(f"{parametro}: {valor}")
    

Mejor_modelo = busqueda_hiperparametros.best_estimator_
y_pred = Mejor_modelo.predict(x_test)

y_proba = Mejor_modelo.predict_proba(x_test)[:,1]


print(f"ROC AUC: {roc_auc_score(y_test, y_proba):.4f}")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")




matriz = confusion_matrix(y_test, y_pred)

print('matriz de confusión:')
print(matriz)



VN, fp, fn ,VF = matriz.ravel()
print(f"Verdaderos Negativos: {VN}")
print(f"Falsos Positivos: {fp}")
print(f"Falsos Negativos: {fn}")
print(f"Verdaderos Positivos: {VF}")



print(classification_report(y_test, y_pred))
```
